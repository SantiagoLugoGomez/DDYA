## 1) Descripción del problema **[1 pts]**
Una plataforma de streaming esta analizando el crecimiento de sus nuevos suscriptores. Se ha observado que el numero de nuevos usuarios registrados en un mes n sigue un comportamiento recursivo determinado por los mese anteriores.

Mes 0: se registraron 0 nuevos usuarios.

Mes 1: se registro un nuevo usuario.

Mes>=2: el numero de nuevos usuarios es igual a la suma de los nuevos usuarios de los dos meses anteriores.

Objetivo: determinar con memorización cuantos nuevos usuarios habra en el mes n.

```python
# Santiago Lugo Gomez codigo parcial practico Tercio 1

def suscripciones_memo(n: int, memo: dict = None) -> int:
    """Calcula el crecimiento de suscriptores para el mes n usando Programacion Dinamica (Top-Down)."""
    if memo is None:
        memo = {}


    if n == 0:
        return 0
    if n == 1:
        return 1


    if n in memo:
        return memo[n]

    memo[n] = suscripciones_memo(n - 1, memo) + suscripciones_memo(n - 2, memo)

    return memo[n]

def main():
    print("=== Calculadora de Crecimiento de Suscriptores ===")

    try:
        n_meses = int(input("Ingrese el numero de meses (n): "))

        if n_meses < 0:
            print("Por favor, ingrese un número entero positivo o cero.")
            return


        memoria = {}


        resultado = suscripciones_memo(n_meses, memoria)


        print("\n--- Resultados ---")
        print(f"Nuevos suscriptores proyectados para el mes {n_meses}: {resultado}")
        print(f"Estado de la memoria despues del calculo:\n{memoria}")

    except ValueError:
        print("Error: Debe ingresar un numero entero valido.")


if __name__ == "__main__":
    main()


```

## 7) Tests aplicados **[15 pts]**

```python
import unittest


# --- Función a probar ---
def suscripciones_memo(n: int, memo: dict = None) -> int:
    if memo is None:
        memo = {}

    if n < 0:
        raise ValueError("El número de meses debe ser mayor o igual a cero.")

    if n == 0:
        return 0
    if n == 1:
        return 1

    if n in memo:
        return memo[n]

    memo[n] = suscripciones_memo(n - 1, memo) + suscripciones_memo(n - 2, memo)
    return memo[n]


# --- Suite de Pruebas Unitarias ---
class TestSuscripcionesMemo(unittest.TestCase):

    def test_casos_base(self):
        """Prueba que los meses 0 y 1 retornen los valores iniciales esperados."""
        self.assertEqual(suscripciones_memo(0), 0)
        self.assertEqual(suscripciones_memo(1), 1)

    def test_valores_intermedios(self):
        """Prueba la secuencia de Fibonacci para valores intermedios conocidos."""
        # n=2 -> 1, n=3 -> 2, n=4 -> 3, n=5 -> 5, n=6 -> 8, n=10 -> 55
        self.assertEqual(suscripciones_memo(2), 1)
        self.assertEqual(suscripciones_memo(3), 2)
        self.assertEqual(suscripciones_memo(4), 3)
        self.assertEqual(suscripciones_memo(5), 5)
        self.assertEqual(suscripciones_memo(6), 8)
        self.assertEqual(suscripciones_memo(10), 55)

    def test_memoizacion_y_cache(self):
        """Verifica que los valores calculados realmente se guarden en el diccionario."""
        memo = {}
        resultado = suscripciones_memo(5, memo)

        self.assertEqual(resultado, 5)
        # Comprobar que los subproblemas n=2, 3, 4, 5 estén guardados en memoria
        self.assertIn(2, memo)
        self.assertIn(3, memo)
        self.assertIn(4, memo)
        self.assertIn(5, memo)
        self.assertEqual(memo[5], 5)

    def test_valor_grande_eficiencia(self):
        """Prueba que un valor grande (ej. n=50) se calcule rápidamente en O(n)."""
        # Con recursión simple sin memoización esto causaría un timeout/crash.
        resultado = suscripciones_memo(50)
        self.assertEqual(resultado, 12586269025)

    def test_excepcion_numero_negativo(self):
        """Verifica que se lance una excepción si se ingresa un mes negativo."""
        with self.assertRaises(ValueError):
            suscripciones_memo(-5)


# --- Ejecutor de pruebas dentro del Notebook o Consola ---
if __name__ == "__main__":
    unittest.main(argv=["first-arg-is-ignored"], exit=False)
```
